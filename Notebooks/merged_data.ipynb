```python
import glob
import os
import pandas as pd
import numpy as np

base_path = "/Users/olaoluwatunmise/Germany-Energy-Price-Forecast/PowerCast-Datasets"

#parent_path = os.path.abspath(base_path)

# Use recursive search inside all subdirectories and  **/ for a recursive search:
csv_file = glob.glob(os.path.join(base_path, "**/*Quarterhour.csv"), recursive=True)

print(csv_file)

for i in csv_file:
    print(i)
```

```python

# Select the 3 specific files
selected_files = [
    "Actual_generation_202301010000_202503050000_Quarterhour.csv",
    "Actual_consumption_202301010000_202503050000_Quarterhour.csv",
    "Cross-border_physical_flows_202301010000_202503050000_Quarterhour.csv"
]


# Filter the files to get full paths
selected_paths = [f for f in csv_file if any(name in f for name in selected_files)]


# Read and merge the files
df_list = [pd.read_csv(file,delimiter=";", low_memory=False) for file in selected_paths]

# Concatenate all DataFrames
merged_df = pd.concat(df_list, ignore_index=True)


# Save to a new CSV file
output_path = "/Users/olaoluwatunmise/Germany-Energy-Price-Forecast/Merged_Quarterhour.csv"
merged_df.to_csv(output_path, index=False)

print(f"Merged CSV saved to: {output_path}")
```

```python
# Adjust display settings for wider output
pd.set_option('display.width', 300)  # Increase the display width
pd.set_option('display.max_columns', None)  # Show all columns
pd.set_option('display.max_rows', 10)  # Limit rows displayed for clarity

df = pd.read_csv(output_path, low_memory=False)
df.head()
```

# Exploratory Data Analysis

```python
# Clean the column names
df.columns = df.columns.str.replace(r'[^\w\s]', '', regex=True)  # Remove special characters
df.columns = df.columns.str.strip()  # Remove leading/trailing spaces
df.columns = df.columns.str.replace('Original resolutions', '', regex=False)  # Remove redundant words
# Remove "MWh" from the column names
df.columns = df.columns.str.replace(' MWh ', '', regex=False)
# Replace spaces between words with underscores for consistency
df.columns = df.columns.str.replace(' ', '_', regex=False)

# Replace multiple consecutive underscores with a single one (in case multiple spaces exist)
df.columns = df.columns.str.replace('_+', '_', regex=True)

# Ensure all column names are lowercase (optional, for consistency)
df.columns = df.columns.str.lower()

df.replace("-", np.nan, inplace=True)  # Ensure '-' is NaN

#new_consumption = consumption

print(df.columns)
print(df.head())
```

```python
# 1. Convert "Start date" to datetime (handle errors)
df["timestamp"] = pd.to_datetime(
    df["start_date"], 
    format="%b %d, %Y %I:%M %p",  # Add explicit format to avoid warnings
    errors='coerce'
)

# Drop "End date" column early
df.drop(columns=["end_date", "start_date"], inplace=True)

# Set datetime column as index FIRST
df.set_index("timestamp", inplace=True)


# Remove commas and convert to float for specified columns
for col in [ 'net_export', 'netherlands_export',
       'netherlands_import', 'switzerland_export', 'switzerland_import',
       'denmark_export', 'denmark_import', 'czech_republic_export',
       'czech_republic_import', 'luxembourg_export', 'luxembourg_import',
       'sweden_export', 'sweden_import', 'austria_export', 'austria_import',
       'france_export', 'france_import', 'poland_export', 'poland_import',
       'norway_export', 'norway_import', 'belgium_export', 'belgium_import',
       'total_grid_load', 'residual_load', 'hydro_pumped_storage', 'biomass',
       'hydropower', 'wind_offshore', 'wind_onshore', 'photovoltaics',
       'other_renewable', 'nuclear', 'lignite', 'hard_coal', 'fossil_gas',
       'other_conventional']:
    # Replace commas and convert to float, assuming it's one number with commas as separators
    df[col] = df[col].str.replace(',', '').astype(float)

#Aggregate Duplicates Before Resampling
df = df.groupby(df.index).mean()  # or .sum(), .max()

# Resample using the DatetimeIndex
df_hourly = df.resample('1h').sum()


# Reset index if needed (optional)
df_hourly.reset_index(inplace=True)

print(df_hourly.head())
```

```python
df_drop = ['netherlands_export',
       'netherlands_import', 'switzerland_export', 'switzerland_import',
       'denmark_export', 'denmark_import', 'czech_republic_export',
       'czech_republic_import', 'luxembourg_export', 'luxembourg_import',
       'sweden_export', 'sweden_import', 'austria_export', 'austria_import',
       'france_export', 'france_import', 'poland_export', 'poland_import',
       'norway_export', 'norway_import', 'belgium_export', 'belgium_import',
       ]

#Create a copy of df and drop the columns
df_filtered = df_hourly.drop(columns=df_drop, errors = 'ignore')

# Check the new DataFrame
print(df_filtered.head())

```
